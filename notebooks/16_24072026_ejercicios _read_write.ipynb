```python
#Ejercicio 01
def extrac_contenido(ruta: str) -> list[str]:
    mi_lista = []

    with open(ruta, "r", encoding="utf-8") as f:
        contenido = f.read()
        lineas = contenido.split("\n")

        for linea in lineas:
            if "lista" in linea.lower():
                continue

            if not linea:
                continue

            if "instrucciones" in linea.lower():
                break

            mi_lista.append(linea)

    return mi_lista


contenido = extrac_contenido("../data/estudiantes_python.txt")

print(contenido)

```

```python
#Ejercicio 02
def extract_contenido(ruta: str) -> list[str]:
    mi_lista = []

    with open(ruta, "r") as f:
        contenido = f.read()
        lineas = contenido.split("\n")

        for linea in lineas:
            if "registro" in linea.lower():
                continue
            if "id," in linea.lower():
                continue
            if not linea:
                continue
            if "instrucciones" in linea.lower():
                break
            mi_lista.append(linea)

    return mi_lista


contenido = extract_contenido("../data/ventas_polos_aqp.txt")


```

```python

class Venta:

    def __init__(
        self,
        id: str,
        producto: str,
        talla: str,
        color: str,
        cliente: str,
        ciudad: str,
        cantidad: int,
        precio_unitario: float,
        metodo_pago: str,
        estado: str
    ) -> None:

        self.id = id
        self.producto = producto
        self.talla = talla
        self.color = color
        self.cliente = cliente
        self.ciudad = ciudad
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario
        self.metodo_pago = metodo_pago
        self.estado = estado
        self.total = cantidad * precio_unitario
```

```python
class Ventas:

    def __init__(self, contenido: list[str]) -> None:
        self.ventas = []

        for linea in contenido:
            datos = linea.split(",")

            venta = Venta(
                datos[0].strip(),
                datos[1].strip(),
                datos[2].strip(),
                datos[3].strip(),
                datos[4].strip(),
                datos[5].strip(),
                int(datos[6].strip()),
                float(datos[7].strip()),
                datos[8].strip(),
                datos[9].strip()
            )

            self.ventas.append(venta)

    def __len__(self) -> int:
        return len(self.ventas)

    def ingreso_total(self) -> float:
        total = 0
        for venta in self.ventas:
            if venta.estado.lower() == "entregado":
                total += venta.total
        return total

    def contar_estados(self) -> dict:
        estados = {"entregado": 0, "pendiente": 0, "cancelado": 0}
        for venta in self.ventas:
            estado = venta.estado.lower()
            if estado in estados:
                estados[estado] += 1
        return estados

    def ventas_arequipa(self) -> int:
        contador = 0
        for venta in self.ventas:
            if venta.ciudad.lower() == "arequipa":
                contador += 1
        return contador

    def metodo_pago_mas_usado(self) -> str:
        metodos = {}
        for venta in self.ventas:
            metodo = venta.metodo_pago
            if metodo in metodos:
                metodos[metodo] += 1
            else:
                metodos[metodo] = 1
        return max(metodos, key=metodos.get)

    def producto_mas_vendido(self) -> str:
        productos = {}
        for venta in self.ventas:
            if venta.producto in productos:
                productos[venta.producto] += venta.cantidad
            else:
                productos[venta.producto] = venta.cantidad
        return max(productos, key=productos.get)

    def resumen_productos(self) -> dict:
        resumen = {}
        for venta in self.ventas:
            if venta.producto not in resumen:
                resumen[venta.producto] = {
                    "cantidad": 0,
                    "ingreso": 0
                }

            resumen[venta.producto]["cantidad"] += venta.cantidad

            if venta.estado.lower() == "entregado":
                resumen[venta.producto]["ingreso"] += venta.total

        return resumen

    def export_resumen(self, ruta: str) -> None:
        with open(ruta, "w") as f:
            f.write("Producto,Cantidad,Ingreso\n")
            resumen = self.resumen_productos()

            for producto, datos in resumen.items():
                f.write(f"{producto},{datos['cantidad']},{datos['ingreso']}\n")

```

```python
ventas = Ventas(contenido)

print(f"Total de ventas: {len(ventas)}")
print(f"Ingreso total: S/ {ventas.ingreso_total():.2f}")

# Estados de las ventas
estados = ventas.contar_estados()
print(f"Entregadas: {estados['entregado']}")
print(f"Pendientes: {estados['pendiente']}")
print(f"Canceladas: {estados['cancelado']}")

# Ventas en Arequipa
print(f"Número de ventas hechas en Arequipa: {ventas.ventas_arequipa()}")

# Método de pago más usado
print(f"Método de pago más usado: {ventas.metodo_pago_mas_usado()}")

# Producto más vendido
print(f"Producto más vendido por cantidad: {ventas.producto_mas_vendido()}")

# Resumen por producto
for producto, datos in ventas.resumen_productos().items():
    print(f"- {producto}: {datos['cantidad']} unidades, S/. {datos['ingreso']:.2f}")

ventas.export_resumen("../data/resumen_ventas.csv")
print("Archivo exportado correctamente.")
```
