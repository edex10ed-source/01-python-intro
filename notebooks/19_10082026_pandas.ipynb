```python
import pandas as pd

s = pd.Series(
    [120, 85, 200, 45],
    index=["ene", "feb", "mar", "abr"],
    name="ventas"
)

print(s["mar"])       # acceso por etiqueta
print(s * 1.21)       # vectorizado: IVA

print(s.mean(), s.max())
```
