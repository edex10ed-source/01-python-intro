####Barras

```python
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D']
valores = [40, 70, 30, 90]

plt.bar(categorias, valores, color='#B628FA')
plt.xlabel('Categoría')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')
plt.show()
```

###Grafico Pastel


```python
import matplotlib.pyplot as plt

frutas = ['Manzana', 'Plátanos', 'Peras']
valores = [30, 50, 20]

plt.pie(valores, labels=frutas, autopct='%1.1f%%')
plt.title('Distribución de frutas')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(1, 20, 100)

dashboard, axes = plt.subplots(2, 4, figsize=(16, 8))

axes[0, 0].plot(x, np.cos(x))
axes[0, 1].plot(x, np.sin(x))
axes[0, 2].plot(x, x)
axes[0, 3].plot(x, x**2)

axes[1, 0].plot(x, -x)
axes[1, 1].plot(x, np.cos(x) * 2)
axes[1, 2].plot(x, np.sin(x) * 2)
axes[1, 3].plot(x, -np.sin(x))

dashboard.tight_layout()
plt.show()
```
