```python
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns

print(sns.__version__)

```

```python
penguins = sns.load_dataset('penguins')
penguins.head(10)
```

```python
sns.histplot(data=penguins, x="body_mass_g", bins=15)

plt.title("Distribución del peso corporal de los pingüinos")
plt.xlabel("Peso corporal (g)")
plt.ylabel("Frecuencia")

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 6 bins
sns.histplot(data=penguins, x="body_mass_g", bins=6, ax=axes[0])
axes[0].set_title("6 bins")
axes[0].set_xlabel("Peso corporal (g)")
axes[0].set_ylabel("Frecuencia")

# 25 bins
sns.histplot(data=penguins, x="body_mass_g", bins=25, ax=axes[1])
axes[1].set_title("25 bins")
axes[1].set_xlabel("Peso corporal (g)")
axes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
sns.histplot(
    data=penguins,
    x="flipper_length_mm",
    bins=18,
    kde=True
)

plt.title("Distribución de la longitud de la aleta")
plt.xlabel("Longitud de la aleta (mm)")
plt.ylabel("Frecuencia")

plt.show()
```

```python
sns.histplot(
    data=penguins,
    x="bill_length_mm",
    hue="species",
    multiple="layer"
)

plt.title("Longitud del pico por especie")
plt.xlabel("Longitud del pico (mm)")
plt.ylabel("Frecuencia")

plt.show()
```
