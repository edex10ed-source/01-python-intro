```python
%sql
-- Ejercicio 1
--Muestra nombre y precio de todos los productos, del más caro al más barato.

SELECT nombre, precio
FROM productos
ORDER BY precio DESC;
```

```python
%sql
-- Ejercicio 2
--Lista los clientes de España o México cuyo nombre empiece por «M».
SELECT *
FROM clientes
WHERE (pais = 'ES' OR pais = 'MX')
  AND nombre LIKE 'M%';
```

```python
%sql
--Ejercicio 3
--Obtén los productos con precio entre 20 y 60 € que tengan stock disponible (> 0)
SELECT *
FROM productos
WHERE precio >= 20
  AND precio <= 60
  AND stock > 0;
```

```python
%sql
--Ejercicio 4
--Muestra los clientes sin teléfono, sustituyendo el NULL por «sin dato» (COALESCE).
SELECT
    nombre,
    COALESCE(telefono, 'sin dato') AS telefono
FROM clientes;
```

```python
%sql
--Ejecicio 5
--Lista los 10 productos más caros con su precio con IVA (× 1.21) usando un alias.
SELECT
    nombre,
    precio,
    precio * 1.21 AS precio_con_iva
FROM productos
ORDER BY precio DESC
LIMIT 10;
```

```python
%sql
-->Ejercicio 6
--Clasifica los productos en "barato", "medio" o "caro" con CASE.
SELECT
    nombre,
    precio,
    CASE
        WHEN precio < 20 THEN 'Barato'
        WHEN precio >= 20 AND precio <= 100 THEN 'Medio'
        ELSE 'Caro'
    END AS categoria_precio
FROM productos;
```

```python
%sql
--Ejercicio 7
--¿Cuántos países distintos hay en la tabla clientes? (DISTINCT)

SELECT COUNT(DISTINCT pais) as total_paises
FROM clientes
```
