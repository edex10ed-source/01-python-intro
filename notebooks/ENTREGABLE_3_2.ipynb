```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.figure import Figure
df=pd.read_csv(r"C:\Users\SpideGamer\Documents\ciencia-de-datos-g8\01-python-intro\data\iris_dataset.csv")
df
```

```python
# 9. Histograma: longitud del sépalo

plt.figure(figsize=(7, 5))

plt.hist(
    df["sepal_length_cm"],
    bins=10,
    edgecolor="black"
)

plt.title("Histograma de la longitud del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Frecuencia")

plt.show()
```

Observación: La mayor parte de las observaciones se concentra en los valores intermedios de la longitud del sépalo, aproximadamente entre 5 y 6,5 cm

```python
# 10. Histograma y bins

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histograma con 5 bins
axes[0].hist(
    df["petal_length_cm"],
    bins=5,
    edgecolor="black"
)

axes[0].set_title("Longitud del pétalo - 5 bins")
axes[0].set_xlabel("Longitud del pétalo (cm)")
axes[0].set_ylabel("Frecuencia")


# Histograma con 20 bins
axes[1].hist(
    df["petal_length_cm"],
    bins=20,
    edgecolor="black"
)

axes[1].set_title("Longitud del pétalo - 20 bins")
axes[1].set_xlabel("Longitud del pétalo (cm)")
axes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

Observación: Con 5 bins, los datos se agrupan en intervalos más grandes y la distribución se observa de forma más general. Con 20 bins, los intervalos son más pequeños y se pueden apreciar más detalles y variaciones en la distribución

```python
# 11. Boxplot por especie

plt.boxplot([
    df[df["species"] == "setosa"]["sepal_length_cm"],
    df[df["species"] == "versicolor"]["sepal_length_cm"],
    df[df["species"] == "virginica"]["sepal_length_cm"]
])

plt.xticks([1, 2, 3], ["setosa", "versicolor", "virginica"])

plt.title("Longitud del sépalo por especie")
plt.xlabel("Especie")
plt.ylabel("Longitud del sépalo (cm)")

plt.show()
```

Observación: Las medianas de la longitud del sépalo son diferentes entre las especies. La especie cuya línea de mediana se encuentra más arriba presenta la mayor longitud mediana del sépalo.

```python
# 12. Boxplot de variables numéricas

plt.boxplot([
    df["sepal_length_cm"],
    df["sepal_width_cm"],
    df["petal_length_cm"],
    df["petal_width_cm"]
])

plt.xticks([1, 2, 3, 4], ["Sepal L", "Sepal W", "Petal L", "Petal W"])

plt.title("Boxplot de variables numéricas")
plt.ylabel("Medidas (cm)")

plt.show()
```

Se observa que las variables tienen diferentes rangos de valores. Las variables de longitud presentan valores mayores que las variables de ancho, por lo que existe una diferencia de escala entre ellas

```python
# 13. Subplots 1x2

plt.figure(figsize=(10, 4))

# Gráfico de la izquierda: scatter
plt.subplot(1, 2, 1)
plt.scatter(df["sepal_length_cm"], df["sepal_width_cm"])
plt.title("Longitud vs. ancho del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Ancho del sépalo (cm)")

# Gráfico de la derecha: histograma
plt.subplot(1, 2, 2)
plt.hist(df["sepal_length_cm"], bins=10, edgecolor="black")
plt.title("Histograma de longitud del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```
